```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#reading the data
data=pd.read_csv("../data/US-pumpkins.csv")
```

```python
print(data.shape)
data.head()
```

```python
#
data=data[['City Name','Package','Date','Low Price','High Price',]]
data.shape
```

```python
avgprice = (data.loc[:,'Low Price'] + data.loc[:,'High Price'])/2
data1=pd.DataFrame({"City": data.iloc[:,0],"Avgprice":avgprice})
data1.City

```

```python
data1.City.value_counts()
```

```python
data1.groupby("City")['Avgprice'].mean().plot(kind='bar')
city=data1.City.unique()
plt.show()
```

```python
x=data1.groupby('City')['Avgprice'].mean()
city=data1.City.unique()
```

```python
sns.catplot(x=x,y=city,kind='box')
```

```python
sns.relplot(x=x,y=city,kind='line')
```
