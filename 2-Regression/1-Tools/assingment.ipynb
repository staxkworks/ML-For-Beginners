```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets,linear_model,model_selection
```

```python
#loading the dataset.
X,y=datasets.load_linnerud(return_X_y=True)
print(X.shape)
print(y.shape)
print(X[0:3])
print(y[0:3])
```

```python
#processing the data
X=X[:,1]
X=X.reshape((-1,1))
print(X.shape)
y=y[:,1]
y=y.reshape((-1,1))
print(y.shape)
```

```python
#splitting the data for testing and training
X_train,X_test,y_train,y_test=model_selection.train_test_split(X,y,test_size=.70)
print(X_test.shape)
print(y_train.shape)
```

```python
#training the model
model=linear_model.LinearRegression()
model.fit(X_train,y_train)
```

```python
#predicting the output
y_pred=model.predict(X_test)
```

```python
#plotting the values
plt.scatter(X_test,y_test,color="blue",)
plt.plot(X_test,y_pred,color="purple",linewidth=2)
plt.xlabel("the waist")
plt.ylabel("no of situps")
plt.title("graph b/w wasitline and situps")
plt.show()
```
