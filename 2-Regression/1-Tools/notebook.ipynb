```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets,linear_model,model_selection
```

```python
#loading thr dataset from the sklear website
# X is the data amd Y is the target value as this is supervised learning.nb
X,y=datasets.load_diabetes(return_X_y=True)
print(X.shape)
print(X[0])
```

```python
#reshaping the data into uselful format as to plot we need two axis
X=X[:,2]
print(X.shape)
X=X.reshape((-1,1))
print(X.shape)
```

```python
#spilitting the data into test data and training data.
X_train,X_test,y_train,y_test=model_selection.train_test_split(X,y,test_size=0.33)
print(X_train.shape)
```

```python
#training the model using the test data
model=linear_model.LinearRegression()
model.fit(X_train,y_train)
```

```python
#using test data to prodecy the new target
y_pred=model.predict(X_test)
```

```python
#plotting the data using matplotlib
plt.scatter(X_test,y_test,color="pink",)
plt.plot(X_test,y_pred,color="red",linewidth=2)
plt.xlabel("Scaled BMI")
plt.ylabel("Disease Progression")
plt.title("A Graph Plot Showing Diabetes Progression Against BMI")
plt.show()
```
